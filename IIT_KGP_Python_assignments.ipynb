```python
##Assignment-1-Write a Python program that computes the following alternating factorial series for a given integer input ip
##function to find factorial of a number
def findfact(n):
  if n<0 :
    return("factorial not supported for negative numbers")
  fact=1
  for i in range(1,n+1):
    fact=fact*i
  return (fact)

##code to check input value is less than or equal to zero
ip=int(input("enter a number"))
if ip<0:
  print("the value of the sum is ",999)
elif ip==0:
  print("the value of the sum is",0)
else:
  #code to calculate the sum defined in the problem
  total_sum = 0.0 # Initialize total_sum
  for k in range(1,ip+1): #loop k till ip
   value= ((-1)**k)*(findfact(k)*k)/(findfact(k)+k**2) #calculate sum
   total_sum = total_sum + value # running total
  print("Final value of value is", round(total_sum,2))
```

```python
# Assignment 1 - Below is the command line version
# ==========================================================
# IIT KHARAGPUR AI4ICPS HUB FOUNDATION
# Hands-on Approach to AI (Cohort 5), September 2026
# Programming Assignment 1
# ==========================================================

import sys

# ----------------------------------------------------------
# Function: frac
# Description: Computes the factorial of a number recursively.
# You are allowed to edit inside this function only.
# ----------------------------------------------------------
def frac(n):
    # TODO: Implement the recursive factorial function and remove the pass statement below.
    if n < 0:
        # As per the problem context, k will always be >= 1 in the loop, so n won't be negative here.
        # However, a robust factorial function should handle this.
        # Returning a string here could cause TypeErrors later, it's better to raise ValueError
        # or return a specific error code like -1 or None, but for this problem, k is always positive.
        return "factorial not supported for negative numbers" # This line might not be reached for k in series
    fact = 1
    for i in range(1, n + 1):
        fact = fact * i
    return fact



# ----------------------------------------------------------
# Function: series
# Description: Computes the alternating factorial series
# S = Σ (-1)^k * (k * frac(k)) / (frac(k) + k^2), for k = 1 to ip
# Special rules:
#   - If ip < 0, return -999.0
#   - If ip == 0, return 0.0
#   - The final result must be rounded to two decimal places
# You are allowed to edit inside this function only.
# ----------------------------------------------------------
def series(ip):
    # TODO: Implement the logic for computing the series and remove the pass statement below.
    # The 'ip' is passed as an argument, so no need for input() here.
    if ip < 0:
        return -999.0
    elif ip == 0:
        return 0.0
    else:
        total_sum = 0.0
        for k in range(1, ip + 1):
            fact_k = frac(k)
            # Ensure frac(k) returns a number. For k >= 1, frac(k) will be an int.
            # Check for division by zero: frac(k) + k**2 should not be zero.
            # For k >= 1, frac(k) >= 1 and k**2 >= 1, so frac(k) + k**2 >= 2, no division by zero.
            term = ((-1)**k) * ((k * fact_k) / (fact_k + k**2))
            total_sum += term
        # The final result must be rounded to two decimal places
        return round(total_sum, 2)


# ----------------------------------------------------------
# Main function: DO NOT MODIFY
# ----------------------------------------------------------
if __name__ == "__main__":
    ip = int(sys.argv[1])
    print(series(ip))
```
